# Graph feature ablation

All comparisons are on the same held-out split. Model selection happened on a separate validation split.

```python
from pathlib import Path
import pandas as pd, matplotlib.pyplot as plt
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
selection = pd.read_csv(ROOT/'reports/model_selection.csv')
test = pd.read_csv(ROOT/'reports/model_comparison_test.csv')
selection, test
```

```python
test.sort_values('pr_auc').plot.barh(x='model', y='pr_auc', title='Held-out PR-AUC'); plt.show()
```
