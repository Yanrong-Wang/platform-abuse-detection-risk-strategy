```python
from pathlib import Path
import pandas as pd, json, matplotlib.pyplot as plt
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
quality = json.loads((ROOT/'reports/data_quality.json').read_text())
relations = pd.read_csv(ROOT/'reports/relation_summary.csv')
quality, relations
```

```python
pd.Series({'normal': quality['nodes']-quality['positive'], 'filtered': quality['positive']}).plot.bar(title='Class balance'); plt.ylabel('Nodes'); plt.show()
```

```python
association = pd.read_csv(ROOT/'reports/feature_association.csv')
association.head(12).sort_values('point_biserial_r').plot.barh(x='feature', y='point_biserial_r', title='Training-only feature association'); plt.show()
```
